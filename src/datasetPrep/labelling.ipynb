This Notebook prepares dataset labels based on 9 expert obstetricians outcomes using majority voting. 

Each obstetrician recorded their evaluation in a csv file. Data is publicly available on: https://people.ciirc.cvut.cz/~spilkjir/data.html


```python
#imports
import pandas as pd
import numpy as np
```

```python
'''
Load the final list of records and their labels (Ground Truth)
This file contains the 'rec_id' needed to link to the .dat files.
'''
file_path = "..\ExpertAnnotations\ExpertAnnCTU-UHB-CTG_20150203 (1).xls"
df = pd.read_csv(file_path)
```

```python
'''
Calculate Majority Vote Function
Function to calculate the mode (most frequent value)
.mode() returns the most frequent value(s). We use [0] to select the first one 
in case of a tie (which is a standard way to handle ties in this context).
'''
def get_majority_vote(series):
    return series.mode()[0]
```

```python
''' Data Focus, one for each step'''
df_votes1 = df[['rec_id', 'eval_step1']].copy()
df_votes2 = df[['rec_id', 'eval_step2']].copy()
df_votes3 = df[['rec_id', 'eval_step3']].copy()
df_votes4 = df[['rec_id', 'eval_step4']].copy()
```

```python
majority_labels1 = df_votes1.groupby('rec_id')['eval_step1'].agg(get_majority_vote).reset_index()
majority_labels1.columns = ['rec_id', 'Majority_Vote_Label']

majority_labels2 = df_votes2.groupby('rec_id')['eval_step2'].agg(get_majority_vote).reset_index()
majority_labels2.columns = ['rec_id', 'Majority_Vote_Label']

majority_labels3 = df_votes3.groupby('rec_id')['eval_step3'].agg(get_majority_vote).reset_index()
majority_labels3.columns = ['rec_id', 'Majority_Vote_Label']

majority_labels4 = df_votes4.groupby('rec_id')['eval_step4'].agg(get_majority_vote).reset_index()
majority_labels4.columns = ['rec_id', 'Majority_Vote_Label']
```

```python
# Include the -1 mapping for completeness, though it shouldn't appear in the final labels.
label_map = {
    1: 'No Hypoxia (Normal)', 
    2: 'Mild Hypoxia (Suspicious)', 
    3: 'Severe Hypoxia (Pathological)',
    -1: 'Uninterpretable (Filtered)'
}
majority_labels1['Clinical_Label'] = majority_labels1['Majority_Vote_Label'].map(label_map)
majority_labels2['Clinical_Label'] = majority_labels2['Majority_Vote_Label'].map(label_map)
majority_labels3['Clinical_Label'] = majority_labels3['Majority_Vote_Label'].map(label_map)
majority_labels4['Clinical_Label'] = majority_labels4['Majority_Vote_Label'].map(label_map)
```

```python
# Display the first few records
print("--- Final Majority Vote Labels for the First 10 Records ---")
print(majority_labels1.head(10).to_markdown(index=False))
```

```python
# Display the class distribution (confirming only 1, 2, or 3 appear)
print("\n--- Distribution of Final Labels Step 1---")
print(majority_labels1['Clinical_Label'].value_counts(normalize=True).to_markdown())

print("\n--- Distribution of Final Labels Step 2---")
print(majority_labels2['Clinical_Label'].value_counts(normalize=True).to_markdown())

print("\n--- Distribution of Final Labels Step 3---")
print(majority_labels3['Clinical_Label'].value_counts(normalize=True).to_markdown())

print("\n--- Distribution of Final Labels Step 4---")
print(majority_labels4['Clinical_Label'].value_counts(normalize=True).to_markdown())
```

```python
# Save the final labels to a new CSV file
majority_labels1.to_csv("../ExpertAnnotations/step1_labels.csv", index=False)
majority_labels2.to_csv("../ExpertAnnotations/step2_labels.csv", index=False)
majority_labels3.to_csv("../ExpertAnnotations/step3_labels.csv", index=False) 
majority_labels4.to_csv("../ExpertAnnotations/step4_labels.csv", index=False)   
print("\nLabels saved to 'ExpertAnnotations/step1_labels.csv', 'ExpertAnnotations/step2_labels.csv', 'ExpertAnnotations/step3_labels.csv', and 'ExpertAnnotations/step4_labels.csv'.")
```
